###  First Chart

```python
import matplotlib.pyplot as plt

colors = ["maroon", "purple", "darkgreen", "orange","darkblue"]

df.groupby("Region")["TotalPrice"].sum().plot(
    kind="bar",
    color=colors
)

plt.title("Region Wise Total Sales")
plt.xlabel("Region")
plt.ylabel("Total Sales")

plt.show()
```
